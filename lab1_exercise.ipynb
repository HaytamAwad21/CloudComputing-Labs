```python
import json

students = [
    {
        "first_name": "Haytam",
        "last_name": "Awad",
        "email": "haytam@e.braude.ac.il",
        "courses": ["Cloud Computing", "Algorithms", "HCI"],
        "link": "https://fittrack1of1.vercel.app"
    },
    {
        "first_name": "Taiseer",
        "last_name": "Ashkar",
        "email": "taiseer@e.braude.ac.il",
        "courses": ["Cloud Computing", "AI"],
        "link": "https://cpstest.org"
    }
]

with open(PATH, 'w', encoding='utf-8') as f:
    json.dump(students, f, ensure_ascii=False, indent=2)
```

```python
import json
import ipywidgets as w
from IPython.display import display

def load():
    with open(PATH, encoding='utf-8') as f:
        return json.load(f)

def save(students):
    with open(PATH, 'w', encoding='utf-8') as f:
        json.dump(students, f, ensure_ascii=False, indent=2)

students = load()

st = {'description_width': '130px'}
L = w.Layout(width='500px')

dd      = w.Dropdown(options=[(f"{s['first_name']} {s['last_name']}", i) for i, s in enumerate(students)],
                     description='Student:', style=st, layout=L)
first   = w.Text(description='First name:', disabled=True, style=st, layout=L)
last    = w.Text(description='Last name:', disabled=True, style=st, layout=L)
email   = w.Text(description='Braude email:', disabled=True, style=st, layout=L)
courses = w.Textarea(description='Courses:', disabled=True, style=st, layout=L)
link    = w.Text(description='Interesting link:', disabled=True, style=st, layout=L)
fav     = w.Text(description='Favorite program:', placeholder='e.g. VS Code', style=st, layout=L)
btn     = w.Button(description='Update', button_style='success', icon='save')
out     = w.Output()

def show(change=None):
    s = students[dd.value]
    first.value   = s['first_name']
    last.value    = s['last_name']
    email.value   = s['email']
    courses.value = '\n'.join(s['courses'])
    link.value    = s['link']
    fav.value     = s.get('favorite_program', '')
    out.clear_output()

def update(b):
    s = students[dd.value]
    s['favorite_program'] = fav.value.strip()
    save(students)
    with out:
        out.clear_output()
        print(f"✅ Saved favorite program for {s['first_name']}: {s['favorite_program']}")

dd.observe(show, names='value')
btn.on_click(update)
show()

header = w.HTML("<h2 style='color:#4a90e2;font-family:sans-serif'>🎓 Group 9 – Students</h2>")
display(w.VBox([header, dd, w.HTML('<hr>'), first, last, email, courses, link,
                w.HTML('<hr>'), fav, btn, out],
               layout=w.Layout(border='2px solid #4a90e2', padding='15px',
                               width='560px', background_color='#f5f9ff')))
```
